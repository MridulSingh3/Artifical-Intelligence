```python
import numpy as np
arr1=np.zeros((2,3))
print(arr1, arr1.shape)
```

```python
arr2=np.ones((5,))
print(arr2, arr2.shape)

arr3=np.full((5,6), 100) # prefill this value
print(arr3, arr3.shape)

arr4=np.eye(3,3) # identity matrix
print(arr4, arr4.shape)

arr5=np.arange(1,11,2) # 1-10
print(arr5, arr5.shape)

arr6=np.linspace(1,100,4) # evenly spaced arrays
print(arr6)
```
