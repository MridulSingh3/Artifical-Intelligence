```python
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset("titanic")
```

```python
df.head() 
```

```python
sns.scatterplot(x="age", y="fare",data=df, palette=["red","green","black"], hue="class")
```

```python
sns.histplot(df["age"],kde=True,bins=25) # bins means number of bars
```

```python
sns.barplot(x="sex", y="fare", data=df, palette=["red","green"])
```

```python
sns.boxplot(x="pclass", y="fare", data=df, palette=["red","green"], hue="sex")
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
```
