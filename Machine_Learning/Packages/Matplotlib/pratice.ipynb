```python
import matplotlib.pyplot as plt
```

```python
#Create a line chart showing the marks of a student in 5 subjects.
subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]
marks = [85, 78, 82, 90, 95]

plt.plot(subjects,marks,marker='o')
plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()
```

```python
#Create a bar chart showing the sales of 5 products.

products = ["A", "B", "C", "D", "E"]
sales = [120, 150, 100, 180, 200]

plt.bar(products,sales)
plt.title("Product Data")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

```python
#Create a scatter plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 8, 10]

plt.scatter(x,y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
#Create a pie chart showing
languages = ["Java", "Python", "JavaScript", "C++", "Other"]
percentage = [30, 25, 20, 15, 10]

plt.pie(percentage, labels=languages, autopct="%1.1f%%")

plt.title("Programming language usage")
plt.show()
```

```python
#Create a histogram using:

import matplotlib.pyplot as plt

marks = [
    45, 55, 67, 72, 81, 90, 56, 78, 88, 95,
    62, 71, 69, 84, 92, 50, 76, 80, 65, 73
]

plt.hist(marks, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Create a line chart and add:

# Title
# X-axis label
# Y-axis label
# Grid

import matplotlib.pyplot as plt

subjects = ["Math", "Physics", "Chemistry", "English", "Computer"]
marks = [85, 78, 82, 90, 95]

plt.plot(subjects, marks, marker="o")

plt.title("Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.grid(True)

plt.show()
```

```python
# Create a bar chart and:

# Add values above each bar
# Give it a title
# Rotate x-axis labels by 45°

products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
sales = [120, 150, 100, 180, 200]

bars = plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.xticks(rotation=45)
    
for i in range(len(products)):
    plt.text(products[i],sales[i]+5,str(sales[i]),ha="center") 

plt.show()
```

```python
# Create a scatter plot of age vs salary:
# age = [21, 22, 23, 24, 25, 26]
# salary = [25000, 28000, 32000, 35000, 42000, 50000]
# Add title and axis labels.

age = [21, 22, 23, 24, 25, 26]
salary = [25000, 28000, 32000, 35000, 42000, 50000]

plt.scatter(age, salary)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

```python
# Create a pie chart of monthly expenses:
# categories = ["Food", "Rent", "Travel", "Shopping", "Other"]
# expenses = [5000, 12000, 3000, 4000, 2000]
# Display percentages on the chart.

categories = ["Food", "Rent", "Travel", "Shopping", "Other"]
expenses = [5000, 12000, 3000, 4000, 2000]

plt.pie(expenses,labels=categories,autopct="%1.1f%%")
plt.title("Monthly Expenses")
plt.show()
```

```python
#Create a histogram of 100 randomly generated exam marks and divide them into 10 bins.
import numpy as np
marks = np.random.randint(0, 101, 100)

plt.hist(marks, bins=10)

plt.title("Distribution of Exam Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Given employee data:
# employees = ["A", "B", "C", "D", "E"]
# salary = [35000, 45000, 30000, 60000, 50000]

# Create a bar chart and highlight the employee with the highest salary.

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 150, 130, 180, 200]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")

# Find highest sales
max_sales = max(sales)
index = sales.index(max_sales)
highest_month = months[index]

print("Highest Sales Month:", highest_month)
print("Highest Sales:", max_sales)

plt.show()
```

```python
# Plot two lines on the same graph:
# months = ["Jan", "Feb", "Mar", "Apr", "May"]
# product_A = [100, 120, 140, 160, 180]
# product_B = [80, 110, 130, 170, 190]

# Add a legend.

months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_A = [100, 120, 140, 160, 180]
product_B = [80, 110, 130, 170, 190]

plt.plot(months, product_A, marker="o", label="Product A")
plt.plot(months, product_B, marker="o", label="Product B")

plt.title("Product Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()

plt.show()
```
