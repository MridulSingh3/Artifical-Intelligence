```python
import matplotlib.pyplot as plt
```

```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]
plt.title("Oscar Movie Revenue Data in Each Year")
plt.xlabel("Year")
plt.ylabel("Revenue in ($M)")
plt.bar(oscar_years,oscar_revenue, color="purple")

for i in range(len(oscar_years)):
    plt.text(oscar_years[i],oscar_revenue[i]+10,str(oscar_revenue[i]),ha="center") # +10 for gap in y axis
```

```python
import numpy as np

oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]


non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

x=np.arange(len(oscar_years))
width=0.4

plt.bar(x-width/2,oscar_revenue, width, label="Oscar Movies")
plt.bar(x+width/2,non_oscar_revenue, width,label="Non-Oscar Movies")

plt.title("Oscar Vs Non-Oscars Movie")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()
plt.xticks(x,oscar_years)
```
