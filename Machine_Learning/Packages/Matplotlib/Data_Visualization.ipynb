```python
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

oscar_years = [2008, 2009, 2010, 2011, 2012]

oscar_revenue = [1005, 170, 427, 133, 232]
plt.title("Oscar Movie Revenue Data in Each Year")
plt.xlabel("Year")
plt.ylabel("Revenue in ($M)")
plt.plot(oscar_years,oscar_revenue)
plt.show()
```

```python
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.plot(oscar_years,oscar_revenue, "o--r" ,label="Oscar_Movie")
plt.plot(oscar_years,non_oscar_revenue, "<--y" ,label="Non_Oscar_Movie")
plt.title("Oscar Movie vs Non Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue in ($M)")
plt.legend()
plt.tight_layout()
```
