```python
import pandas as pd

data = {
    "Name": ["Rahul", "Aman", "Priya", "Neha", "Rohit", "Anjali", "Vikas", "Sneha"],
    "Age": [21, 23, 20, 22, 24, 21, 25, 22],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Salary": [45000, 35000, 50000, 42000, 60000, 38000, 55000, 48000],
    "Experience": [1, 2, 1, 3, 4, 1, 5, 2],
    "City": ["Delhi", "Agra", "Mathura", "Delhi", "Agra", "Mathura", "Delhi", "Agra"]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Display the first 5 rows.
print(df.head())

#Display the last 3 rows.
print()
print(df.tail(3))
```

```python
#Find the number of rows and columns.
print(df.shape)

#Display all column names.
print(df.columns)
```

```python
#Display the data types of all columns.
print(df.dtypes)

#Find basic statistical information using describe().
print()
print(df.describe())
```

```python
#Display only the Name and Salary columns.
print(df[["Name","Salary"]])

#Display the Name, Department, and City columns.
print(df[["Name","Department","City"]])
```

```python
#Find the average salary.
print(df["Salary"].mean())

#Find the maximum and minimum salary.
print(df["Salary"].max())
print(df["Salary"].min())
```

```python
#Find employees whose salary is greater than 50000.
print(df[df["Salary"]>50000])

#Find employees whose age is greater than 21.
print()
print(df[df["Age"]>21])
```

```python
#Find employees who belong to the IT department.
print(df[df["Department"]=='IT'])

#Find employees from Agra.
print()
print(df[df["City"]=="Agra"])
```

```python
#Find employees with salary between 40000 and 55000.
print(df[(df["Salary"]>=40000) & (df["Salary"]<=55000)])

#Find employees who have more than 2 years of experience.
print()
print(df[(df["Experience"]>2)])
```

```python
#Find employees who are from Agra and work in the IT department.
print(df[(df["City"]=="Agra") & (df["Department"]=="IT")])

#Find employees who are from either Delhi or Mathura.
print()
print(df[(df["City"]=="Delhi") | (df["City"]=="Mathura")])
```

```python
#Sort employees by salary in ascending order.
df=df.sort_values("Salary",ascending=True)
print(df)

#Sort employees by salary in descending order.
print()
df=df.sort_values("Salary",ascending=False)
print(df)
```

```python
#Sort employees by Department and then by Salary.
df=df.sort_values(["Department","Salary"])
print(df)

#Add a new column called Bonus equal to 10% of salary.
print()
df["Bonus"]=df["Salary"]*0.10
print(df)
```

```python
#Add a column TotalSalary:
df["TotalSalary"]=df["Salary"]+df["Bonus"]
print(df)

#Increase everyone's salary by 5%.
print()
df["Salary"]=df["Salary"]*1.05
print(df)
```

```python
#Rename the Name column to Employee_Name.
df=df.rename(columns={"Name":"Employee_Name"})
print(df)
```

```python
#Find the average salary of each department.
print(df.groupby("Department")["Salary"].mean())

#Find the maximum salary in each department.
print()
print(df.groupby("Department")["Salary"].max())

#Find the minimum salary in each department.
print()
print(df.groupby("Department")["Salary"].min())

#Count number of employees in each department
print()
print(df.groupby("Department")["Salary"].count())
```

```python
# Find the average salary of employees from each city.
print(df.groupby("City")["Salary"].mean())

# Find the total salary paid by each department.
print()
print(df.groupby("Department")["Salary"].sum())

# Find the employee count for each city.
print()
print(df.groupby("Department")["Salary"].count())
```

```python
data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Rohit"],
    "Age": [21, None, 23, 22, None],
    "Salary": [40000, 50000, None, 45000, 60000],
    "City": ["Agra", "Delhi", None, "Mathura", "Agra"]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Find the number of missing values in each column.
print(df.isna().sum())

# Remove rows containing missing values.
df=df.dropna()
print(df)
```

```python
# Replace missing Age values with the average age.
avg_age = df["Age"].mean()
df["Age"] = df["Age"].fillna(avg_age)
print(df)


# Replace missing Salary values with the average salary.
print()
avg_salary=df["Salary"].mean()
df["Salary"]=df["Salary"].fillna(avg_salary)
print(df)

#Replace missing City with "Unknown"
print()
df["City"]=df["City"].fillna("Unknown")
print(df)
```

```python
#Find the second-highest salary
second_highest=df["Salary"].drop_duplicates().nlargest(2).iloc[-1]
print(second_highest)

#Find the employee with the highest salary
print()
highest_salary=df["Salary"].nlargest(1)
print(highest_salary)
```

```python
#Find the employee with the highest salary
print(df.loc[df["Salary"].idxmax()])

#Find the highest-paid employee in each department
print()
result = df.loc[df.groupby("Department")["Salary"].idxmax()]
print(result)
```

```python
#Departments having average salary greater than 45000
result = df.groupby("Department")["Salary"].mean()
print(result[result > 45000])
```
